```python
import speech_recognition as sr
from transformers import pipeline
from pydub import AudioSegment
```

```python
recognizer = sr.Recognizer()
qa_pipeline = pipeline('question-answering')
```

```python
mp3_file = '/content/class.mp3'
audio = AudioSegment.from_mp3(mp3_file)
wav_file = '/content/demo.wav'
audio.export(wav_file, format='wav')
```

```python
with sr.AudioFile(wav_file) as source:
    audio_data = recognizer.record(source)
    text = recognizer.recognize_google(audio_data)
    print("Transcribed text:", text)
```

```python
context_text = text
language_models = {
    "French": "Helsinki-NLP/opus-mt-en-fr",
    "German": "Helsinki-NLP/opus-mt-en-de",
    "Spanish": "Helsinki-NLP/opus-mt-en-es",
    "Hindi": "Helsinki-NLP/opus-mt-en-hi",
    "Chinese": "Helsinki-NLP/opus-mt-en-zh",
    "Arabic": "Helsinki-NLP/opus-mt-en-ar",
    "Russian": "Helsinki-NLP/opus-mt-en-ru",
    "Japanese": "Helsinki-NLP/opus-mt-en-ja",
    "Korean": "Helsinki-NLP/opus-mt-en-ko",
    "Italian": "Helsinki-NLP/opus-mt-en-it",
    "Portuguese": "Helsinki-NLP/opus-mt-en-pt",
    "Bengali": "Helsinki-NLP/opus-mt-en-bn",
    "Tamil": "Helsinki-NLP/opus-mt-en-ta",
    "Telugu": "Helsinki-NLP/opus-mt-en-te",
    "Marathi": "Helsinki-NLP/opus-mt-en-mr",
    "Urdu": "Helsinki-NLP/opus-mt-en-ur"
}

translation_pipelines = {}
for lang, model_name in language_models.items():
    try:
        translation_pipelines[lang] = pipeline('translation', model=model_name)
    except Exception as e:
        print(f"Warning: Could not load translation model for {lang}. Error: {e}")

while True:
    question = input("Enter your question (or 'quit' to exit): ")
    if question.lower() == 'quit':
        break
    try:
        qa_result = qa_pipeline(question=question, context=context_text)
        answer = qa_result['answer']
        print(f"Answer in English: {answer}")
    except Exception as e:
        print("Error in question answering:", e)
        continue
    for lang, translator in translation_pipelines.items():
        try:
            translated = translator(answer, max_length=100)[0]['translation_text']
            print(f"{lang}: {translated}")
        except Exception:
            print(f"{lang}: [Translation failed]")
    print("-" * 40)
```
